find the average and highest marks

```python
students={
    "Anu":85,
    "Rahul":72,
    "Meera":91,
    "Arun":68
}
total=0
highest=0
top_student=""
for name,mark in students.items():
    total+=mark
    if mark>highest:
        highest=mark
        top_student=name
average=total/len(students)
print("Average:",average)
print("Top student:",top_student)
print("Highest Mark:",highest)
```

function + Loop + List

```python
def get_even_numbers(numbers):
    even_numbers=[]
    for num in numbers:
        if num%2==0:
            even_numbers.append(num)
    return even_numbers

numbers=[10,15,20,23,30,35]
print(get_even_numbers(numbers))
```

```python
def highest_student(students):
    highest = 0
    name_of_student = ""

    for name, mark in students.items():
        if mark > highest:
            highest = mark
            name_of_student = name

    return name_of_student, highest


students = {
    "Anu": 85,
    "Rahul": 72,
    "Meera": 91
}

name, mark = highest_student(students)

print("Student:", name)
print("Mark:", mark)
```
